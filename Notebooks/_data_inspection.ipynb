```python
#Load and First Look at the Data

import pandas as pd 
import numpy as np

df=pd.read_csv('../DATA/Raw/ai4i2020.csv')
print("Shape", df.shape)
display(df.head(5))
```

```python

print("--- Data Info ---")
df.info()

print("\n--- Missing Values ---")
print(df.isnull().sum())

print("\n--- Duplicate Rows ---")
print(df.duplicated().sum())
```

```python
# Overall failure indicator
print("--- Machine Failure (Overall) ---")
print(df['Machine failure'].value_counts())
print(f"Failure rate: "
      f"{df['Machine failure'].mean()*100:.2f}%")

# 5 specific failure types
failure_types = ['TWF', 'HDF', 'PWF', 'OSF', 'RNF']

print("\n--- Individual Failure Type Counts ---")
for col in failure_types:
    count = df[col].sum()
    pct = df[col].mean() * 100
    print(f"{col}: {count} occurrences ({pct:.2f}%)")

# Check overlap — can multiple failures happen at once?
print("\n--- Failure Overlap Check ---")
df['failure_count'] = df[failure_types].sum(axis=1)
print(df['failure_count'].value_counts().sort_index())
```

```python
sensor_cols = ['Air temperature [K]',
               'Process temperature [K]',
               'Rotational speed [rpm]',
               'Torque [Nm]',
               'Tool wear [min]']

print("--- Sensor Reading Summary ---")
display(df[sensor_cols].describe().round(2))

print("\n--- Product Type Distribution ---")
print(df['Type'].value_counts())
```
