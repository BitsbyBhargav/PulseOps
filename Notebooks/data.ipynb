```python
import pandas as pd

df = pd.read_csv('../data/raw/ai4i2020.csv')

df['temp_diff'] = (df['Process temperature [K]'] 
                   - df['Air temperature [K]'])

df['low_temp_diff_flag'] = (
    df['temp_diff'] < 8.6).astype(int)

df['high_torque_flag'] = (
    df['Torque [Nm]'] > 
    df['Torque [Nm]'].quantile(0.75)
    ).astype(int)

df['high_tool_wear_flag'] = (
    df['Tool wear [min]'] > 
    df['Tool wear [min]'].quantile(0.75)
    ).astype(int)

df.to_csv('../data/processed/pulseops_enriched.csv', index=False)
print("Done:", df.shape)
```
